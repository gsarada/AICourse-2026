```python
from bing_create.main import ImageGenerator
import os

# --- Configuration ---
# Replace 'YOUR_SRCHHPGUSR_COOKIE' with the cookie value you copied
cookie_u = 'cookie_u'
COOKIE = 'cookie_SRCHLANG'

def create_image(prompt, OUTPUT_DIR = 'content/generated_images'):

  # Create output directory if it doesn't exist
  if not os.path.exists(OUTPUT_DIR):
      os.makedirs(OUTPUT_DIR)

  # Initialize the BingCreate object
  # This tool uses reverse engineering of the Bing Image Creator API
  try:
      generator = ImageGenerator(auth_cookie_u= cookie_u, auth_cookie_srchhpgusr=COOKIE)

      print(f"Generating images for prompt: '{prompt}'...")

      # Generate images (default is 4 images)
      images = generator.generate(prompt=prompt, num_images=1)

      # Save the images to the directory
      generator.save(
          images=images,
          output_dir=OUTPUT_DIR
      )

      print("Image generation and saving complete.")

  except Exception as e:
      print(f"An error occurred: {e}")
      print("Please check if your SRCHHPGUSR cookie is valid and you are signed in to Bing Image Creator.")
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

image_dir = 'content/generated_images'
def print_images(image_dir):
  image_files = [f for f in os.listdir(image_dir) if f.endswith(('.jpg', '.jpeg'))]
  image_files.sort() # Sort to maintain a consistent order

  if not image_files:
      print(f"No images found in {image_dir}")
  else:
      print(f"Displaying images from {image_dir}:")
      plt.figure(figsize=(20, 15))
      for i, image_file in enumerate(image_files):
          img_path = os.path.join(image_dir, image_file)
          img = mpimg.imread(img_path)

          plt.subplot(2, (len(image_files) + 1) // 2, i + 1) # Adjust subplot grid dynamically
          plt.imshow(img)
          plt.title(image_file)
          plt.axis('off')
      plt.tight_layout()
      plt.show()

print_images(image_dir)
```

## Task 2: Image Comparison Observations

Based on the newly displayed images, please document your observations for each pair:

### Robot (Problematic vs. Solution Prompts)

**Problematic Prompt: 'robot dancing in a nightclub'**
- **Unintentional Elements:** This prompt generated a literal interpretation of a robot dancing in a nightclub, leading to generic imagery. The directness of the prompt resulted in predictable, uninspired compositions.

**Solution Prompt: 'humanoid figure in fluid motion, dark environment with colourful spotlights, dynamic pose, metallic surface reflections'**

- **Cliche Avoidance (how it worked):** This prompt successfully avoided the cliche by describing the desired visual outcome without using the problematic trigger words 'robot' and 'nightclub'. It guided the AI to create an image that evokes the feeling of a dancing robot in a nightclub but with a more unique and sophisticated aesthetic.

### Magical Forest (Problematic vs. Solution Prompts)

**Problematic Prompt: 'magical forest with fairies'**
- **Unintentional Elements:** This prompt typically led to stereotypical representations of fairies and magical forests, featuring common fairy designs (e.g., small, winged humanoids) and saccharine aesthetics. The inclusion of 'fairies' as a direct keyword resulted in a predictable, fairy-tale-like image that lacks originality.

**Solution Prompt: 'misty woodland at dawn, small points of bioluminescent light between trees, ethereal atmosphere, soft bokeh effect'**

- **Cliche Avoidance (how it worked):** By replacing 'fairies' with descriptions of light, atmosphere, and natural elements, the prompt successfully generated an image that conveys a sense of wonder and magic without resorting to the conventional and often clichéd depiction of fairies. The result is a more unique and atmospheric magical forest.
