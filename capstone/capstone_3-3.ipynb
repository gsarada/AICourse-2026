```python
from bing_create.main import ImageGenerator
import os

# --- Configuration ---
# Replace 'YOUR_SRCHHPGUSR_COOKIE' with the cookie value you copied
cookie_u = 'cookie_u'
COOKIE = 'cookie_SRCHLANG'

def create_image(prompt, OUTPUT_DIR = 'content/generated_images'):

  # Create output directory if it doesn't exist
  if not os.path.exists(OUTPUT_DIR):
      os.makedirs(OUTPUT_DIR)

  # Initialize the BingCreate object
  # This tool uses reverse engineering of the Bing Image Creator API
  try:
      generator = ImageGenerator(auth_cookie_u= cookie_u, auth_cookie_srchhpgusr=COOKIE)

      print(f"Generating images for prompt: '{prompt}'...")

      # Generate images (default is 4 images)
      images = generator.generate(prompt=prompt, num_images=1)

      # Save the images to the directory
      generator.save(
          images=images,
          output_dir=OUTPUT_DIR
      )

      print("Image generation and saving complete.")

  except Exception as e:
      print(f"An error occurred: {e}")
      print("Please check if your SRCHHPGUSR cookie is valid and you are signed in to Bing Image Creator.")
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

def print_images(image_dir):
  image_files = [f for f in os.listdir(image_dir) if f.endswith(('.jpg', '.jpeg'))]
  image_files.sort() # Sort to maintain a consistent order

  if not image_files:
      print(f"No images found in {image_dir}")
  else:
      print(f"Displaying images from {image_dir}:")
      plt.figure(figsize=(10, 10))
      for i, image_file in enumerate(image_files):
          img_path = os.path.join(image_dir, image_file)
          img = mpimg.imread(img_path)

          plt.subplot(2, (len(image_files) + 1) // 2, i + 1) # Adjust subplot grid dynamically
          plt.imshow(img)
          plt.title(image_file)
          plt.axis('off')
      plt.tight_layout()
      plt.show()
```

```python
#Part 1: Technical Specifications Impact
prompts = {'sample1': {'base': 'portrait of a young woman with blonde hair', 'technical': 'portrait of a young woman with blonde hair, shot on Hasselblad H6D, 80mm lens at f/2.8, natural window light, Kodak Portra 400 film simulation, shallow depth of field'},
           'sample2': {'base': 'modern city street at night', 'technical': 'modern city street at night, Sony A7R IV, 24mm f/1.4 lens, long exposure, high contrast, neon reflections on wet pavement, 8K resolution'}}
for sample in prompts:
  base_prompt = prompts[sample]['base']
  technical_prompt = prompts[sample]['technical']
  create_image(base_prompt, OUTPUT_DIR = 'content/generated_images/task1')
  create_image(technical_prompt, OUTPUT_DIR = 'content/generated_images/task1')
```

### Optimal Prompt Structures for Image Generation

Optimal prompt structures often follow a logical flow, moving from general concepts to specific details, and leveraging keywords and modifiers to guide the AI effectively.

1.  **Start with the Core Subject/Concept:** Begin with the main subject of your image. This is the primary focus.
    *   *Example:* "A majestic lion", "A bustling city street"

2.  **Define the Action or Interaction (if any):** If your subject is doing something, describe that next.
    *   *Example:* "A majestic lion **roaring at sunset**", "A bustling city street **with people and vehicles**"

3.  **Specify the Environment/Setting:** Describe where the scene is taking place.
    *   *Example:* "A majestic lion roaring at sunset **in the African savanna**", "A bustling city street with people and vehicles **at night in Tokyo**"

4.  **Add Artistic Style and Medium:** Clearly state the desired artistic style, genre, or medium. This is crucial for guiding the aesthetic.
    *   *Example:* "A majestic lion roaring at sunset in the African savanna, **oil painting style**", "A bustling city street with people and vehicles at night in Tokyo, **cinematic photograph**"

5.  **Incorporate Technical/Photographic Details:** This is where you add camera specifics, lighting, composition, and other technical modifiers.
    *   *Example:* "A majestic lion roaring at sunset in the African savanna, oil painting style, **warm golden light, dramatic brushstrokes**"
    *   *Example:* "A bustling city street with people and vehicles at night in Tokyo, cinematic photograph, **long exposure, neon reflections, 8K resolution, f/1.4 lens**"

6.  **Refine with Mood, Emotion, and Additional Descriptors:** Add terms that evoke the desired atmosphere, feeling, or specific visual characteristics.
    *   *Example:* "A majestic lion roaring at sunset in the African savanna, oil painting style, warm golden light, dramatic brushstrokes, **powerful, untamed, vibrant colors**"
    *   *Example:* "A bustling city street with people and vehicles at night in Tokyo, cinematic photograph, long exposure, neon reflections, 8K resolution, f/1.4 lens, **futuristic, rainy, high contrast**"

**General Structure Template:**

`[Subject] [Action/Interaction], [Environment/Setting], [Artistic Style/Medium], [Technical/Photographic Details], [Mood/Emotion/Additional Descriptors]`

**Key Considerations for Optimal Structures:**

*   **Clarity and Conciseness:** Use clear, unambiguous language. Avoid jargon unless it's a specific technical term you intend to use.
*   **Keyword Density:** Ensure important keywords are present, but don't overstuff the prompt; focus on quality over quantity.
*   **Comma Separation:** Use commas to delineate distinct ideas or clauses, helping the AI parse different elements of your prompt.
*   **Prioritization (Implicit):** While not explicitly stated with weights, the order of elements often implies importance. What comes first tends to be weighted more heavily by the AI.
*   **Iterative Approach:** Optimal structures are often found through experimentation. Start with a simpler version and gradually add complexity, observing how each addition influences the output.

### Best Practices for Writing Image Generation Prompts

1.  **Be Specific and Descriptive:** Provide as much detail as possible about the subject, style, mood, and environment. Instead of "a dog", try "a fluffy golden retriever puppy playing in a sun-drenched meadow."
2.  **Use Technical Terms Judiciously:** As seen in our experiments, specific camera models, lens types, lighting setups (e.g., "Hasselblad H6D", "80mm lens at f/2.8", "split lighting setup"), and artistic styles (e.g., "Impressionist style", "matte painting") can significantly influence the output.
3.  **Specify Lighting:** Lighting is crucial for mood and realism. Use terms like "natural light", "golden hour sunlight", "studio strobe", "neon reflections", or describe the quality (e.g., "soft shadows", "harsh light").
4.  **Define Style and Medium:** Clearly state the desired artistic style or medium. Examples include "oil painting", "digital art", "photograph", "anime", "cinematic", "hyperrealistic."
5.  **Control Composition and Perspective:** Include terms related to framing (e.g., "portrait", "wide shot", "close-up"), angle (e.g., "low angle", "bird's eye view"), and depth of field (e.g., "shallow depth of field").
6.  **Specify Resolution/Quality:** Terms like "4K concept art", "8K resolution", or "high detail" can push the generator for higher fidelity.
7.  **Consider Mood and Emotion:** Describe the desired feeling or atmosphere (e.g., "calm and serene", "vibrant", "moody", "sad").
8.  **Avoid Negations (when possible):** It's often more effective to state what you want rather than what you don't want. For example, instead of "not blurry", try "sharp focus."
9.  **Iterate and Experiment:** Image generation is an iterative process. Start with a simpler prompt and gradually add more detail and technical terms. Observe the changes and refine your prompts accordingly.
10. **Use Commas and Clear Language:** Separate distinct ideas with commas to help the AI parse your prompt effectively. Use clear, unambiguous language.
11. **Leverage Examples:** If you have a specific visual in mind, describe elements from similar images you've seen.
