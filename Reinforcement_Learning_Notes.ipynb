# Reinforcement Learning (RL)
 Is a machine learning paradigm where an agent learns to make decisions by interacting with an environment to maximize cumulative rewards. Key concepts include the agent performing actions (A), observing states (S), receiving numerical rewards (R), and following a policy (
) to optimize behavior over time.

Agent: The learner or decision-maker that interacts with the environment.
Environment: The world or simulation the agent operates within.

State: A specific, current situation or configuration of the environment.

Action: Moves or decisions the agent can take, forming an "action space" (discrete or continuous).

Reward: Scalar feedback signal from the environment indicating the success of an action.

Policy: The agent's strategy; a mapping from states to actions that determines behavior.

Discount Factor: A factor between 0 and 1 that determines the importance of future rewards, prioritizing immediate rewards as
 approaches 0.

Value Function: Predicts the expected long-term cumulative reward from a given state under a specific policy.

Markov Decision Process (MDP): The formal mathematical framework for modeling RL problems, defining states, actions, transitions, and rewards.

Exploration vs. Exploitation: The tradeoff between trying new actions to discover better strategies (exploration) and choosing known actions to maximize rewards (exploitation).

RL differs from supervised learning because it learns from delayed feedback and experience rather than explicit, labeled examples.

# Expected Value (Value Function)
In RL, the agent aims to maximize long-term rewards, not just immediate gains. The Expected Value (V) of a state represents how "good" it is to be in that state, calculating the average cumulative future reward an agent can expect, usually discounted to prioritize sooner rewards.

Key Concept: Instead of just looking at the next reward, the agent looks at the "expected return" (G<sub>t</sub>).

Formula: V<sup>π</sup>(s)=E<sub>π</sub>[G<sub>t</sub>|S<sub>t</sub>=s] (The expected return starting from state following policy π).

Example: Grid World Maze

Imagine a robot in a 3*3 grid, trying to reach a "+10" reward (goal) while avoiding a "-10" penalty (fire).

State A (next to goal): The agent knows that from here, a "move right" action has a 90% chance of reaching the goal. Therefore, State A has a high Expected Value.

State B (next to fire): The agent knows "move up" has a 50% chance of falling into fire. State B has a low or negative Expected Value.

Result: The agent moves to states with higher Expected Values.

# Markov Property (Markov Decision Process - MDP)
The Markov Property states that the future state depends only on the current state and action, not on the sequence of events that preceded it. In other words, "the future is independent of the past, given the present". This allows the agent to make decisions without knowing the entire history of its actions.

Key Concept: The current state S<sub>t</sub> fully summarizes the necessary information.
Formalization: P(S<sub>t+1</sub>|S<sub>t</sub>) = P(S<sub>t+1</sub>|S<sub>t</sub>,S<sub>t-1</sub>...S<sub>0</sub>)

Example: Chess or Tic-Tac-Toe

Markov Example: When playing chess, the current configuration of pieces on the board (state) tells you everything you need to know to make the next move. It doesn't matter if you reached this position in 10 moves or 50 moves.

Non-Markov Counter-Example: Trying to predict stock prices only based on today's price. You actually need the past 30 days to make an accurate prediction, meaning the current state is not enough.

# Policy(π)
A Policy is the agent's brain or strategy. It is a mapping from the current state (s) to the action (a) that the agent should take. <br>
- Deterministic Policy (π(s)=a): In state s, always take action a.<br>
- Stochastic Policy (π(a|s)= P(A<sub>t</sub> = a | S<sub>t</sub> = s)): In state s, take action a with a certain probability.

Example: Autonomous Vehicle<br>
State: The car detects a pedestrian 50 meters ahead and the traffic light is green.

Policy:<br>
If speed > 40 km/h -> Action: Brake.<br>
If speed < 40 km/h -> Action: Maintain speed.

The goal of training is to find the Optimal Policy (π<sup>*</sup>) that maximizes the long-term expected reward.

#Difference between State Value and Expected Value
In reinforcement learning, the state-value function (V<sup>π</sup>(s)) is a specific type of expected value (specifically, the expected discounted return), but they are not generally synonymous in all contexts.

Here is the breakdown of how they relate:<br>
State-Value  V<sup>π</sup>(s): Represents the expected long-term return (sum of future rewards) an agent will receive starting from state s and following a specific policy π thereafter. It is the expected value of the return, conditioned on being in state s.

Expected Value (General): In probability, this is a weighted average of all possible outcomes. In RL, "expected value" is often used to describe the calculation within the Bellman equation, which averages over possible next states and actions.

Markov Property: The robot's position (2,2) is enough to know it can move Up, Down, Left, or Right. It doesn't need to remember how it got to (2,2).
